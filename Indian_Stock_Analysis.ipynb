```python
import pandas as pd
import yfinance as yf
import datetime
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
s = datetime.datetime(2023,1,1)
e = datetime.datetime(2025,5,1)
```

```python
wipro = yf.download('WIPRO.NS',start=s,end=e)
tcs = yf.download('TCS.NS',start=s,end=e)
irfc = yf.download('IRFC.NS',start=s,end=e)
irctc = yf.download('IRCTC.NS',start=s,end=e)
mrf = yf.download('MRF.NS',start=s,end=e)
hdfc = yf.download('HDFCBANK.NS',start=s,end=e)
```

```python
wipro.columns = wipro.columns.get_level_values(0)
tcs.columns = tcs.columns.get_level_values(0)
irfc.columns = irfc.columns.get_level_values(0)
irctc.columns = irctc.columns.get_level_values(0)
mrf.columns = mrf.columns.get_level_values(0)
hdfc.columns = hdfc.columns.get_level_values(0)
```

```python
wipro = wipro.reset_index()
tcs = tcs.reset_index()
irfc = irfc.reset_index()
irctc = irctc.reset_index()
mrf = mrf.reset_index()
hdfc = hdfc.reset_index()
```

```python
wipro['Symbol'] = 'WIPRO'
tcs['Symbol'] = 'TCS'
mrf['Symbol'] = 'MRF'
irfc['Symbol'] = 'IRFC'
irctc['Symbol'] = 'IRCTC'
hdfc['Symbol'] = 'HDFC'
```

```python
df = pd.concat([tcs,wipro,irctc,irfc,mrf,hdfc],axis=0)
```

```python
print(df.Symbol.unique())
st = input("Enter Stock Name: ")
stk = df[df.Symbol == st]

sb.lineplot(x=stk.Date,y=stk.Close)
plt.xticks(rotation=90)
plt.show()
```
